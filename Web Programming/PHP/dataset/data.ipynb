```python
import matplotlib.pyplot as plt

# Data extracted from the table
models = ["YOLOv11", "Faster R-CNN", "YOLOv6-L", "VGG-19", "RF + ROI", "CNN", "YOLOv12n (Proposed)"]
accuracy = [90.0, 91.66, 92.9, 88.0, 87.0, 89.0, 93.3]

# Different colors for each bar
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd", "#8c564b", "#e377c2"]

# Plotting with smaller bar width
plt.figure(figsize=(10, 6))
bars = plt.bar(models, accuracy, color=colors, edgecolor="black", width=0.5)

# Add value labels on bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.3, f"{yval:.2f}%",
             ha="center", va="bottom", fontsize=10)

plt.title("Model Accuracy Comparison for Brain Tumor Detection", fontsize=14)
plt.xlabel("Models", fontsize=12)
plt.ylabel("Accuracy (%)", fontsize=12)
plt.ylim(85, 95)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```
