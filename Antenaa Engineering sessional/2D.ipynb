```python
# Dipole Antenna-2D 
import numpy as np
import matplotlib.pyplot as plt

# Angle values (0 to 180 degree)
theta = np.linspace(0, np.pi, 360)

# Radiation pattern
R = np.abs(np.sin(theta))

# Polar plot
plt.figure()
ax = plt.subplot(111, projection='polar')
ax.plot(theta, R)

ax.set_title("2D Radiation Pattern of Dipole Antenna", va='top')
plt.show()
```

```python
# Dipole Antenna-3D 
import numpy as np
import matplotlib.pyplot as plt

# Angular values
theta = np.linspace(0, np.pi, 100)
phi = np.linspace(0, 2*np.pi, 100)
theta, phi = np.meshgrid(theta, phi)

# Radiation pattern formula for half-wave dipole
R = np.abs(np.sin(theta))

# Convert to Cartesian coordinates
X = R * np.sin(theta) * np.cos(phi)
Y = R * np.sin(theta) * np.sin(phi)
Z = R * np.cos(theta)

# Plotting 3D pattern
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_title("3D Radiation Pattern of Dipole Antenna")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

```python
# 3D 
import numpy as np
import matplotlib.pyplot as plt

# Angular values
theta = np.linspace(0, np.pi, 100)
phi = np.linspace(0, 2*np.pi, 100)
theta, phi = np.meshgrid(theta, phi)

# Radiation pattern formula for half-wave dipole
R = np.abs(np.sin(theta))

# Convert to Cartesian coordinates
X = R * np.sin(theta) * np.cos(phi)
Y = R * np.sin(theta) * np.sin(phi)
Z = R * np.cos(theta)

# Plotting 3D pattern
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# --- colored surface (smooth 3-color map) ---
surf = ax.plot_surface(X, Y, Z,
                       cmap="plasma",     # try: 'viridis', 'turbo', 'inferno'
                       edgecolor='none')

# color bar (optional)
fig.colorbar(surf, ax=ax, shrink=0.6)

# axis labels with different colors
ax.set_xlabel("X", color="red")
ax.set_ylabel("Y", color="green")
ax.set_zlabel("Z", color="blue")

ax.set_title("3D Radiation Pattern of Dipole Antenna")

plt.show()
```

```python
# Isotropic Antenna- 2D 
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 400)
R = np.ones_like(theta)

ax = plt.subplot(111, projection="polar")
ax.plot(theta, R)
ax.set_title("Isotropic Antenna (2D)")
plt.show()
```

```python
# Isotropic-3D
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, np.pi, 200)
phi = np.linspace(0, 2*np.pi, 200)
theta, phi = np.meshgrid(theta, phi)

R = np.abs(np.cos(theta))

X = R * np.sin(theta) * np.cos(phi)
Y = R * np.sin(theta) * np.sin(phi)
Z = R * np.cos(theta)

ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap="plasma")
ax.set_title("Patch Antenna (3D)")
plt.show()
```

```python
# Horn Antenna-2D
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(-np.pi/2, np.pi/2, 400)
n = 6
R = np.abs(np.cos(theta))**n

ax = plt.subplot(111, projection="polar")
ax.plot(theta, R)
ax.set_title("Horn Antenna (2D)")
plt.show()
```

```python
#Horn Antenna-3D
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, np.pi, 250)
phi = np.linspace(0, 2*np.pi, 250)
theta, phi = np.meshgrid(theta, phi)

n = 6
R = np.abs(np.cos(theta))**n

X = R * np.sin(theta) * np.cos(phi)
Y = R * np.sin(theta) * np.sin(phi)
Z = R * np.cos(theta)

ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap="inferno")
ax.set_title("Horn Antenna (3D)")
plt.show()
```

```python
# Linear Antenna-2D
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, np.pi, 800)
N = 8
d = 0.5
k = 2*np.pi

num = np.sin(N * k * d * np.cos(theta) / 2)
den = N * np.sin(k * d * np.cos(theta) / 2)
AF = np.abs(num / den)

ax = plt.subplot(111, projection="polar")
ax.plot(theta, AF)
ax.set_title("Linear Antenna Array (2D)")
plt.show()
```

```python
# Linear Antenna-3D
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, np.pi, 200)
phi = np.linspace(0, 2*np.pi, 200)
theta, phi = np.meshgrid(theta, phi)

N = 8
d = 0.5
k = 2*np.pi

num = np.sin(N * k * d * np.cos(theta) / 2)
den = N * np.sin(k * d * np.cos(theta) / 2)
R = np.abs(num / den)

X = R * np.sin(theta) * np.cos(phi)
Y = R * np.sin(theta) * np.sin(phi)
Z = R * np.cos(theta)

ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap="coolwarm")
ax.set_title("Linear Antenna Array (3D)")
plt.show()
```
